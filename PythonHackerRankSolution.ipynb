```python
# Problem 6 : Write a function
def is_leap(year):
    leap = False

    # Write your logic here
    if year < 100 :
        if year%4==0 :
            leap = True
    elif year >= 100 :
        if year%400==0:
            leap = True
        elif year%4==0 and year%100!=0:
            leap = True
    return leap

year = int(input())
print(is_leap(year))
```

```python
# Problem 11 : Second Maximum
if __name__ == '__main__':
    n = int(input())
    arr = map(int, input().split())

    res = []
    for i in arr:
        res.append(i)

    print(res)
    mx = float('-inf')
    for i in range(n):
        if res[i] > mx:
            mx = res[i]
    print(mx)
    smx = float('-inf')
    for i in range(n):
        if res[i] != mx and res[i] > smx :
            smx = res[i]
            print(smx)
    print(smx)
```

```python
# Problem 12 : Finding the percentage
if __name__ == '__main__':
    n = int(input())
    student_marks = {}
    for _ in range(n):
        name, *line = input().split()
        scores = list(map(float, line))
        student_marks[name] = scores
    query_name = input()

    li = student_marks[query_name]
    lng = len(li)
    s = sum(li)
    res = s/lng
    res = round(res,2)
    print(f"{res:.2f}")
```

```python
# Problem 13 : Nested Lists
if __name__ == '__main__':
    student = []
    scr = []
    for _ in range(int(input())):
        name = input()
        score = float(input())
        student.append([name,score])
        scr.append(score)

    sorted_scr = sorted(list(set(scr)))
    smn = sorted_scr[1]
    res = []
    for name,score in student:
        if score==smn:
            res.append(name)
    res.sort()
    for i in res:
        print(i)
```

```python
# Problem 18 : Word Capitalize
def solve(s):
  words = s.split(' ')
  res = [word.capitalize() for word in words]
  return res
st = input()
word = solve(st)
print(word)
```
