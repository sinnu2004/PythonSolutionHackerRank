```python
# Problem 2 : Python If-Else
n = int(input("Enter the number : ").strip())
if n%2!=0:
    print("Weird")
elif n%2==0 and n>=2 and n<=5 :
    print("Not Weird")
elif n%2==0 and n>=6 and n<=20 :
    print("Weird")
elif n%2==0 and n>20:
    print("Not Weird")
```

```python
# Problem 6 : Write a function
def is_leap(year):
    leap = False

    # Write your logic here
    if year < 100 :
        if year%4==0 :
            leap = True
    elif year >= 100 :
        if year%400==0:
            leap = True
        elif year%4==0 and year%100!=0:
            leap = True
    return leap

year = int(input())
print(is_leap(year))
```

```python
# Problem 7 : Sum and Prod
import numpy as np
n,m = map(int,input().split())
grid = [list(map(int,input().split())) for _ in range(n)]
my_array = np.array(grid)
sum_0 = np.sum(my_array,axis=0)
prod = np.prod(sum_0)
print(prod)
```

```python
# Problem 8 : Min and Max
import numpy as np

n,m = map(int,input().split())
grid = [list(map(int,input().split())) for _ in range(n)]

arr = np.array(grid)

mini = np.min(arr,axis=1)
maxi = np.max(mini)

print(maxi)
```

```python
# Problem 9:  Mean, Var, and Std
import numpy as np
n,m = map(int,input().split())

grid = [input().split() for _ in range(n)]
arr = np.array(grid,int)

m = np.mean(arr,axis=1)
v = np.var(arr,axis=0)
st = np.std(arr,axis=None)

print(m)
print(v)
print(round(st,11))
```
